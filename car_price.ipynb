```python
import pandas as pd
df = pd.read_csv("data/car_data.csv")
df.head()
```

```python
print(df.shape)
df.info()
```

```python
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().reset_index(drop=True)

for col in ["mileage", "engine", "max_power"]:
    df[col] = df[col].astype(str).str.extract(r"([\d.]+)")[0].astype(float)

df = df.drop(columns=["torque"], errors="ignore")
```

```python
for col in ["mileage", "engine", "max_power", "seats"]:
    df[col] = df[col].fillna(df[col].median())

print(df.isnull().sum())
```

```python
df["car_age"] = 2026 - df["year"]
df["brand"] = df["name"].str.split().str[0]

df = df.drop(columns=["name", "year"])
df.head()
```

```python
df.to_csv("data/cleaned_car_data.csv", index=False)
print(df.shape)
```
